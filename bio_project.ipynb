```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('framingham.csv')
```

```python
data.head()
```

```python
data.shape
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
data.fillna(data[['education','cigsPerDay','BPMeds','totChol','BMI','heartRate','glucose']].mean(),inplace=True)
```

```python
data.isna().sum()
```

```python
data.describe()
```

```python
data.corr()
```

```python
sns.heatmap(data.corr(),annot = True)
```

```python
sns.pairplot(data=data)
```

```python
X=data.drop(columns=['TenYearCHD'])
y=data['TenYearCHD']
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
xcolumns = X.columns
```

```python
X = scaler.fit_transform(X)
```

```python
X = pd.DataFrame(X, columns = xcolumns)
X.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=100)
```

```python
X_train.shape
```

```python
X_test.shape
```

LOGISTIC Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
```

```python
param={'C': [0.01, 0.1, 1, 10],'penalty': ['l1', 'l2'],'solver': ['liblinear']}
```

```python
r=LogisticRegression(max_iter=1000)
```

```python
grid = GridSearchCV(
    r,
    param,
    scoring='recall',
    cv=5
)
```

```python
grid.fit(X_train,y_train)
```

```python
s=cross_val_score(grid,X,y,cv=5,scoring='accuracy')
s.mean()
```

```python
y_pred=grid.predict(X_test)
```

```python
a=metrics.accuracy_score(y_test,y_pred)
a
```

```python
p_log=metrics.precision_score(y_test,y_pred)
p_log
```

```python
metrics.recall_score(y_test,y_pred)
```

```python
metrics.confusion_matrix(y_test,y_pred)
```

Random forest

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
params={'max_depth': [0.5, 1, 5, 10],'n_estimators': [16, 32, 50, 100], 'n_jobs': [1, 2],'random_state': [1, 10, 20, 42]}
```

```python
model=GridSearchCV(RandomForestClassifier(),params,cv=5,scoring='accuracy')
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
b=metrics.accuracy_score(y_test,y_pred)
b
```

```python
metrics.f1_score(y_test,y_pred)
```

```python
p_rand=metrics.precision_score(y_test,y_pred)
p_rand
```

```python
p_r=metrics.recall_score(y_test,y_pred)
```

```python
p_r
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
param_grid={'n_neighbors':[3,31],'weights':['uniform','distance'],'metric':['minkowski' ]}
```

```python
grid_kn=GridSearchCV(KNeighborsClassifier(), param_grid, cv=5)
```

```python
grid_kn.fit(X_train,y_train)
```

```python
grid_kn.best_params_
```

```python
y_pred=grid_kn.predict(X_test)
```

```python
c=metrics.accuracy_score(y_pred,y_test)
c
```

```python
p_k=metrics.precision_score(y_pred,y_test)
p_k
```

SVC

```python

from sklearn.svm import SVC
param={'C': [0.75, 0.85, 0.95, 1], 'degree': [3, 4, 5],'kernel': ['linear', 'poly', 'rbf', 'sigmoid']}
```

```python
R=GridSearchCV(SVC(kernel='rbf', probability=True, random_state=42),param,cv=5,scoring='accuracy')
```
